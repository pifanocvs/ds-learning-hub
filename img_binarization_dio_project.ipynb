### Desafio de Projeto

Seguindo o exemplo do algoritmo de binarização apresentado em nossa última aula, realize a implementação em Python para transformar uma imagem colorida para níveis de cinza (0 a 255) e para binarizada (0 e 255), preto e branco.  

# Imagem Original
Monumento_da_Laranja_Campo_Grande.jpg

# Projeto


```python
import cv2
import numpy as np

#sugestão de teste da lib do cv2
import sys

# sugestão do snippet
from google.colab.patches import cv2_imshow
```

```python
# Leitura da imagem
img = cv2.imread('/Monumento_da_Laranja_Campo_Grande.jpg')

# Teste leitura da imagem
if img is None:
    sys.exit("Could not read the image.")

# *pra ler precisei colocar no ambiente do colab
```

```python
# Conversão para escala de cinza
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Aplicação de blur
suave = cv2.GaussianBlur(img, (7, 7), 0)

# Aplicação de limiarização binária
(T, bin) = cv2.threshold(suave, 160, 255, cv2.THRESH_BINARY)

# Aplicação de limiarização binária invertida
(T, binI) = cv2.threshold(suave, 160, 255, cv2.THRESH_BINARY_INV)

# Combinação das imagens para visualização
resultado = np.vstack([
    np.hstack([suave, bin]),
    np.hstack([binI, cv2.bitwise_and(img, img, mask=binI)])
])
```

```python
# Exibição do resultado - corrigido com base no alerta de snippet do proprio collab
cv2_imshow(resultado)
```
